## Part 2

```python
import numpy as np

def newton(f, df, x0, epsilon=1e-8, max_iter=30):
    x = x0
    n_iter = 0

    while n_iter < max_iter:
        if abs(f(x)) < epsilon:
            print(f"Found root in {n_iter} iterations")
            return x

        x = x - f(x) / df(x)
        n_iter += 1

    print("Iteration failed")
    return None
```

```python
def f(x):
    return x**3 - x**2 - 1

def df(x):
    return 3*x**2 - 2*x

newton(f, df, 1)
```
